```python
import pandas as pd

# Load dataset
df = pd.read_csv(r"D:\Self_project\Health_Care\Healthcare_analysis_Project\dataset\healthcare_dataset.csv")
# Create a copy
privacy_df = df.copy()

# Columns that may directly identify a person
pii_columns = [
    "Name",
    "Patient Name",
    "Patient ID",
    "Phone Number",
    "Mobile Number",
    "Email",
    "Address",
    "Street Address",
    "Aadhaar Number",
    "Passport Number"
]

# Find PII columns
found_pii = []

for column in pii_columns:

    if column in privacy_df.columns:
        found_pii.append(column)

# Remove PII
if len(found_pii) > 0:

    privacy_df.drop(
        columns=found_pii,
        inplace=True
    )

    print("PII removed:", found_pii)

else:

    print("No specified PII columns found.")

# Remove duplicate records
duplicate_count = privacy_df.duplicated().sum()

privacy_df.drop_duplicates(
    inplace=True
)

# Check remaining PII
remaining_pii = []

for column in pii_columns:

    if column in privacy_df.columns:
        remaining_pii.append(column)

# Final privacy check
if len(remaining_pii) == 0:

    print("Privacy check PASSED.")

else:

    print("Privacy check FAILED.")
    print("PII still present:", remaining_pii)

# Save privacy-safe dataset
privacy_df.to_csv(
    "healthcare_privacy_safe.csv",
    index=False
)

print("Privacy-safe dataset saved successfully.")

# Ethical warning
print("""
Ethical Notice:
This dataset should be anonymized and handled securely.
Real patient information should not be uploaded to GitHub.
Model predictions are for educational purposes and should
not be treated as medical diagnosis.
""")
```
